```python
import cv2
import cvzone 
import math
import pygame
import serial
from ultralytics import YOLO
```

```python
model=YOLO(r"C:\Users\yuossef\Desktop\Deif Alla 2\AI\fire.pt")
cap=cv2.VideoCapture(r"C:\Users\yuossef\Desktop\Deif Alla 2\AI\video fire.mp4")
#ser=serial.Serial(COM5,9600)
```

```python
while True:
    ret,frame=cap.read()
    frame=cv2.resize(frame,(640,480))
    result=model(frame, stream=True)
    for i in result:
        boxes=i.boxes
        for box in boxes:
            confidence=box.conf[0]
            confidence=math.ceil(confidence *100)
            if confidence >60:
                x1,y1,x2,y2=boxes.xyxy[0]
                x1,y1,x2,y2=int(x1),int(y1),int(x2),int(y2)
                cv2.rectangle(frame, (x1,y1),(x2,y2),(0,0,255),9)
                cvzone.putTextRect(frame, f'Fire {confidence}%', [x1 + 8, y1 + 100],
                                   scale=1.5, thickness=2)
                #ser.write(b'1')
            #else:
                #ser.write(b'0')
    cv2.imshow("meshadshad", frame)
cap.release()
cv2.destroyAllWindows()


cap.release()
cv2.destroyAllWindows()
```
